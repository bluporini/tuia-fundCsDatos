### 1. Análisis exploratorio de datos

Sean:

- $X$: tiempo de exposición (minutos, segundos, etc.).
- $Y$: número de bacterias sobrevivientes.

Lo primero sería usar estadísticos descriptivos:

```
df.describe()
```

para observar medias, desvíos estándar, mínimos y máximos. Luego, usamos gráfico de dispersión:

```
import matplotlib.pyplot as plt

plt.scatter(df["tiempo"], df["bacterias"])
plt.xlabel("Tiempo de exposición")
plt.ylabel("Bacterias sobrevivientes")
plt.show()
```

Este gráfico es fundamental porque permite visualizar la relación entre ambas variables.

### 2. Relación esperada entre las variables

Desde el punto de vista biológico, al aumentar el tiempo de exposición a 180°C se espera que disminuya la cantidad de bacterias sobrevivientes.

La relación suele ser de tipo exponencial:

$$Y=Ae^{−kX}$$

donde:

- $A$ representa la cantidad inicial de bacterias.
- $k>0$ es la tasa de destrucción bacteriana.

Por lo tanto, el gráfico de dispersión probablemente muestre una curva decreciente y no una recta.

### 3. ¿Puede ajustarse una regresión lineal simple?

Para responder esta pregunta, primero debe analizarse el diagrama de dispersión entre el tiempo de exposición y la cantidad de bacterias sobrevivientes.

Si el gráfico muestra una tendencia aproximadamente lineal decreciente, resulta razonable ajustar un modelo de regresión lineal simple:

$$Y = \beta_0 + \beta_1X + \varepsilon$$

donde:

- $Y$ representa la cantidad de bacterias sobrevivientes.
- $X$ representa el tiempo de exposición.
- $\beta_0$ es la cantidad esperada de bacterias cuando el tiempo de exposición es cero.
- $\beta_1$ mide la variación promedio en la cantidad de bacterias por cada unidad adicional de tiempo.
- $\varepsilon$ representa el error aleatorio.

Dado que se espera que la cantidad de bacterias disminuya al aumentar el tiempo de exposición, se espera obtener un coeficiente $\beta_1<0$.

Si el diagrama de dispersión no muestra una relación aproximadamente lineal, el modelo lineal simple podría no ser adecuado para describir correctamente la relación entre las variables.

### 4. Ajuste del modelo

Una vez observada una tendencia aproximadamente lineal en el gráfico de dispersión, se ajusta el modelo:

```
import statsmodels.api as sm

X = df["tiempo"]
y = df["bacterias"]

X = sm.add_constant(X)

modelo = sm.OLS(y, X).fit()

print(modelo.summary())
```

También puede graficarse la recta ajustada:

```
import matplotlib.pyplot as plt

plt.scatter(df["tiempo"], df["bacterias"])
plt.plot(df["tiempo"], modelo.predict(X), color="red")
plt.xlabel("Tiempo de exposición")
plt.ylabel("Bacterias sobrevivientes")
plt.show()
```

### 5. Interpretación de los coeficientes

Supongamos que el modelo ajustado es:

$$Y = 1200 − 45X$$

Entonces 

- El intercepto $1200$ representa la cantidad estimada de bacterias cuando el tiempo de exposición es cero.
- La pendiente $−45$ indica que por cada unidad adicional de tiempo de exposición, la cantidad de bacterias disminuye en promedio $45$ unidades.

El signo negativo de la pendiente es coherente con el fenómeno estudiado.

### 6. Coeficiente de determinación $R^2$

El coeficiente de determinación mide qué proporción de la variabilidad de la variable respuesta es explicada por el modelo. Por ejemplo, si $R^2 = 0.85$ entonces el 85% de la variabilidad observada en la cantidad de bacterias sobrevivientes es explicada por el tiempo de exposición. Cuanto más cercano a $1$ sea $R^2$|, mejor es el ajuste.

### 7. Análisis de residuos

Para validar el modelo es necesario analizar los residuos. Primero, veamos el gráfico residuos vs ajustados

```
plt.scatter(modelo.fittedvalues, modelo.resid)
plt.axhline(0, color="red")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.show()
```

Se espera observar:

- residuos distribuidos aleatoriamente alrededor de cero;
- ausencia de patrones;
- varianza aproximadamente constante.

También podemos graficar el histograma de residuos:

```
plt.hist(modelo.resid)
plt.show()
```

Permite evaluar si la distribución de los errores es aproximadamente normal.

### 8. Conclusiones

El análisis exploratorio permitió observar una relación negativa entre el tiempo de exposición y la cantidad de bacterias sobrevivientes. El diagrama de dispersión mostró una tendencia aproximadamente lineal decreciente, por lo que se consideró adecuado ajustar un modelo de regresión lineal simple.

El coeficiente asociado al tiempo de exposición resultó negativo, indicando que a medida que aumenta el tiempo de exposición disminuye la cantidad de bacterias sobrevivientes. El valor de $R^2$ mostró que una proporción importante de la variabilidad observada puede explicarse mediante el modelo.

El análisis de residuos no evidenció patrones importantes, por lo que los supuestos del modelo pueden considerarse razonablemente satisfechos. En consecuencia, el modelo ajustado resulta útil para describir y predecir la cantidad de bacterias sobrevivientes en función del tiempo de exposición.
