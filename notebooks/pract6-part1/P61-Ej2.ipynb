### 2. Relación observada entre las variables

Desde un punto de vista socioeconómico es razonable esperar que los países con mayores niveles educativos presenten mejores condiciones sanitarias, mayor acceso a servicios de salud y una mejor calidad de vida.

Por ello suele observarse una relación positiva entre ambas variables:

A mayor educación $\Rightarrow$ Mayor esperanza de vida

Si el gráfico de dispersión muestra una tendencia aproximadamente lineal creciente, entonces resulta razonable plantear un modelo de regresión lineal simple.

### 3. Modelo propuesto

El modelo lineal simple es

$$Y= \beta_0 + \beta_1 X + \varepsilon$$

donde:

- $Y$: esperanza de vida.
- $X$: años de educación.
- $\beta_0$: esperanza de vida esperada cuando los años de educación son cero.
- $\beta_1$: incremento promedio de la esperanza de vida por cada año adicional de educación.
- $\varepsilon$: término de error aleatorio.

### 4. Ajuste del modelo

Utilizando Statsmodels:

```
import statsmodels.api as sm

X = sm.add_constant(df["educacion"])
y = df["esperanza_vida"]

modelo = sm.OLS(y, X).fit()

print(modelo.summary())
```

También puede visualizarse la recta ajustada:

```
plt.scatter(df["educacion"],
            df["esperanza_vida"])

plt.plot(df["educacion"],
         modelo.predict(X),
         color="red")

plt.xlabel("Años de educación")
plt.ylabel("Esperanza de vida")
plt.show()
```


### 5. Interpretación de los coeficientes

Supongamos que el modelo ajustado resulta

$$Y = 55.2 + 1.8X$$

- Intercepto $\hat{β}_0 = 55.2$. Representa la esperanza de vida estimada para una población con cero años de educación. Aunque puede carecer de interpretación práctica, es necesario para definir la recta de regresión.

- Pendiente $\hat{β}_1 = 1.8$. Indica que, en promedio, cada año adicional de educación se asocia con un incremento de 1.8 años en la esperanza de vida.

El signo positivo confirma la relación directa entre ambas variables.

### 6. Coeficiente de determinación $R^2$

El $R^2$ mide la proporción de variabilidad explicada por el modelo. Por ejemplo: $$R^2 = 0.72$$ significa que el 72% de la variabilidad observada en la esperanza de vida puede explicarse mediante los años de educación.

En general:

- $R^2 < 0.3$: relación débil.
- $0.3<R^2<0.7$: relación moderada.
- $R^2>0.7$: relación fuerte.
